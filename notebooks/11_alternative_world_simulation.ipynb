5. **S4 Macro Shock**: Severe inflation (+20% freight) and carrier disruption (+4 days delay).
6. **S5 Combined Blitz**: Dual intervention (-15% freight, -3 days delay).

```python
import sys
from pathlib import Path

REPO_ROOT = Path('.').resolve().parent if Path('.').resolve().name == 'notebooks' else Path('.').resolve()
sys.path.insert(0, str(REPO_ROOT / 'src'))

import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Image, display

# Load Scenario Simulation Results
results_path = REPO_ROOT / 'artifacts' / 'reports' / 'scenario_simulation_results.json'
with open(results_path, 'r') as f:
    sim_data = json.load(f)

print(f"Loaded {len(sim_data)} Counterfactual Policy Scenarios (12-Week Horizon).")
```

## 1. Scenario Comparison Table (Deltas vs S0 Baseline)
We inspect the projected cumulative trade-offs between customer acquisition, satisfaction improvement, and financial sustainability.

```python
summary_rows = []
for s_id, s in sim_data.items():
    cmp = s.get('comparison_to_s0', {})
    summary_rows.append({
        'Scenario ID': s_id,
        'Name': s['name'],
        'Cumul Volume': int(round(s['cumulative_volume'])),
        'Delta Orders': int(round(cmp.get('delta_volume_orders', 0))),
        'Delta Orders (%)': f"{cmp.get('delta_volume_pct', 0):+.2f}%",
        'Review Rate (%)': f"{s['avg_negative_review_rate']*100:.2f}%",
        'Delta Review (pp)': f"{cmp.get('delta_negative_review_pp', 0):+.2f} pp",
        'Subsidy Cost (BRL)': f"R$ {s['subsidy_cost_brl']:,.2f}",
        'Net Profit (BRL)': f"R$ {cmp.get('net_incremental_profit_brl', 0):,.2f}",
        'Guardrail': 'PASS' if not s['guardrail']['is_out_of_distribution'] else 'OOD WARN',
    })

summary_df = pd.DataFrame(summary_rows)
display(summary_df)
```
